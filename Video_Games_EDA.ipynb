DATA VISUALISATIONS:

1. Number of games by genre:

```python
plt.figure(figsize=(10, 6))

top_genres = df['Genre'].value_counts().head(10)
sns.barplot(x=top_genres.values, y=top_genres.index, palette='viridis')

plt.title('Top 10 Game Genre Combinations', fontsize=14, fontweight='bold')
plt.xlabel('Count', fontsize=12)
plt.ylabel('Genre String', fontsize=12)
plt.tight_layout()
plt.show()
```

2. Genre distribution:

```python
plt.figure(figsize=(8, 8))

genre_counts = df['Genre'].value_counts()
top_10 = genre_counts.head(10)
other_sum = pd.Series({'Other': genre_counts.iloc[10:].sum()})
pie_data = pd.concat([top_10, other_sum])

plt.pie(
    pie_data,
    labels=pie_data.index,
    autopct='%1.1f%%',
    startangle=140,
    colors=sns.color_palette('tab10')
)
plt.title('Genre Distribution Share', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

3. Distribution of Votes:

```python
plt.figure(figsize=(8,5))

sns.histplot(df['Votes'], bins=30, kde=True, color='skyblue', edgecolor='black')

plt.title('Distribution of User Votes', fontsize=14, fontweight='bold')
plt.xlabel('Votes', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.tight_layout()
plt.show()
```

4. Votes by Genre:

```python
plt.figure(figsize=(8, 6))

top_10_genre_names = df['Genre'].value_counts().head(10).index
df_top_genres = df[df['Genre'].isin(top_10_genre_names)]

sns.boxplot(
    data=df_top_genres,
    x='Votes',
    y='Genre',
    palette='Set2'
)
plt.title('Votes Distribution across Top 10 Genre Combinations', fontsize=14, fontweight='bold')
plt.xlabel('Votes', fontsize=12)
plt.ylabel('Genre Combination', fontsize=12)
plt.tight_layout()
plt.show()
```

5. Ranking vs Votes:

```python
plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=df,
    x='Ranking',
    y='Votes',
    alpha=0.7,
    color='purple'
)
plt.title('Ranking vs. Votes', fontsize=14, fontweight='bold')
plt.xlabel('Ranking', fontsize=12)
plt.ylabel('Votes', fontsize=12)
plt.tight_layout()
plt.show()
```

6. Top 10 Games by Vote:

```python
plt.figure(figsize=(8, 5))

top10_games = df.sort_values(by='Votes', ascending=False).head(10)

sns.barplot(
    data=top10_games,
    x='Votes',
    y='Name',
    palette='Blues_r'
)
plt.title('Top 10 Most Voted Video Games', fontsize=14, fontweight='bold')
plt.xlabel('Total Votes', fontsize=12)
plt.ylabel('Game Title', fontsize=12)
plt.tight_layout()
plt.show()
```
