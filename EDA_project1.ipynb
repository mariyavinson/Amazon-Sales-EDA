Data Visualizations

Plot 1 - Rating Distribution

```python
plt.figure(figsize=(8,5))
sns.histplot(df['rating'], bins=20, kde=True)
plt.title('Distribution of Customer Ratings')
plt.xlabel('Average Rating (out of 5)')
plt.ylabel('Number of Products')
plt.show()
```

Plot 2 - Price Distribution

```python
plt.figure(figsize=(8,5))
sns.histplot(df['discounted_price'], bins=30, kde=True)
plt.title('Distribution of Discounted Price')
plt.xlabel('Discounted Price (₹)')
plt.ylabel('Number of Products')
plt.show()
```

Plot 3 - Discount percentage

```python
plt.figure(figsize=(8,5))
sns.histplot(df['discount_percentage'], bins=20, kde=True)
plt.title('Distribution of Discount Offered')
plt.xlabel('Discount Percentage (%)')
plt.ylabel('Number of Products')
plt.show()
```

Plot 4 - Category bar chart

```python
plt.figure(figsize=(10,5))
df['main_category'].value_counts().plot(kind='bar')
plt.title('Number of Products by Main Category')
plt.xlabel('Category')
plt.ylabel('Number of Products')
plt.xticks(rotation=45, ha='right')
plt.show()
```

Plot 5 - Price vs Rating

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='discounted_price', y='rating', data=df)
plt.title(' Discounted Price vs Rating')
plt.xlabel('Discounted Price (₹)')
plt.ylabel('Rating')
plt.show()
```

Plot 6 - Outlier Detection

```python
fig, axes = plt.subplots(1, 3, figsize=(15,5))

sns.boxplot(y=df['discounted_price'], ax=axes[0])
axes[0].set_xlabel('Discounted Price')
axes[0].set_ylabel('Price (₹)')

sns.boxplot(y=df['actual_price'], ax=axes[1])
axes[1].set_xlabel('Actual Price')
axes[1].set_ylabel('Price (₹)')

sns.boxplot(y=df['rating_count'], ax=axes[2])
axes[2].set_xlabel('Rating Count')
axes[2].set_ylabel('Number of Ratings')

plt.tight_layout()
plt.show()
```

Plot 7 - Rating by Category

```python
plt.figure(figsize=(12,6))
sns.boxplot(x='main_category', y='rating', data=df)
plt.xticks(rotation=45, ha='right')
plt.title('Rating Distribution by Category')
plt.xlabel('Category')
plt.ylabel('Rating')
plt.show()
```

Plot 8 - Top 10 most-reviewed products

```python
top10 = df.nlargest(10, 'rating_count')
plt.figure(figsize=(10,6))
sns.barplot(x='rating_count', y='product_name', data=top10)
plt.title('Top 10 Most Reviewed Products')
plt.xlabel('Number of Ratings')
plt.ylabel('Product Name')
plt.show()
```
